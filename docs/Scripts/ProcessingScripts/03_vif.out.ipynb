## VIF calculation per item

The VIF (Variance Inflation Factor) diagnoses multicollinearity among items. Values above 5 indicate problematic multicollinearity, anticipating convergence and interpretation issues in CFA.

```

response <- seq_len(nrow(items))
model <- lm(response ~ ., data = items)
vif_values <- car::vif(model)
```

### VIF Table

```

vif_df <- tibble(
  Item = names(vif_values),
  VIF  = round(as.numeric(vif_values), 3),
  Problematic = if_else(VIF > 5, "Yes", "No")
)

vif_df |>
  knitr::kable(
    caption = "Variance Inflation Factor (VIF) per item — VIF > 5 indicates problematic multicollinearity",
    digits  = 3
  )
```
